```python
import requests
import os
import pandas as pd

from pathlib import Path
from ebmdatalab import bq

DATA_DIR = "../data"

os.environ["EBMDATALAB_BQ_CREDENTIALS_PATH"] = 'bq-service-account.json'
```

```python
query = """
SELECT 
  ddd.year_month,
  ddd.vmp_code,
  ddd.ods_code,
  ddd.vmp_name,
  ddd.uom,
  ddd.uom_name,
  ddd.quantity,
  ddd.ddd_quantity,
  ddd.ddd_value,
  ddd.ddd_unit,
  ddd.calculation_logic,
  ddd.ingredient_code,
  vmp.vtm_code,
  vmp.vtm_name
FROM `ebmdatalab.scmd_pipeline.ddd_quantity` ddd
JOIN `ebmdatalab.scmd_pipeline.vmp_data` vmp 
  ON ddd.vmp_code = vmp.vmp_code
WHERE vmp.vtm_code IN ('774399006', '777047009', '777435007')
ORDER BY ddd.year_month, ddd.vmp_code, ddd.ods_code;
"""

df = bq.cached_read(sql=query, csv_path=Path(DATA_DIR, f"cdk46_op_hosp.csv"), use_cache=True)
```
